## 1. Петухи и поросята


x — количество петухов;

y — количество поросят.


```python
x = Int("chikens")
y = Int("pigs")
solver = Solver() # создаём объект-решатель
solver.add(x >= 0, y >= 0, x + y == 11,2 * x + 4 * y == 30)
if solver.check() == sat:
    model = solver.model()
    print("Петухов:", model[x])
    print("Поросят:", model[y])
```

## 2. Задача на логику

**Условие:**

В коробке лежат карандаши: **7 красных** и **5 синих**.
Мальчик достаёт карандаши в темноте (наощупь). Какое **наименьшее** количество карандашей ему нужно взять,
чтобы среди них точно был хотя бы один **красный**?

### модель

В худшем случае мальчик сначала достанет все 5 синих карандашей.
Поэтому для гарантии хотя бы одного красного нужно взять больше 5 карандашей:

`pencils_taken > 5`.

Найдём минимальное целое количество карандашей с помощью `Optimize`.

```python
pencils_taken = Int("pencils_taken")

pencil_solver = Optimize()
pencil_solver.add(
    pencils_taken >= 1,
    pencils_taken <= 12,
    pencils_taken > 5
)
pencil_solver.minimize(pencils_taken)

if pencil_solver.check() == sat:
    pencil_model = pencil_solver.model()
    print("Минимальное количество карандашей:", pencil_model[pencils_taken])
else:
    print("Решение не найдено")
```

Ожидаемый результат: нужно взять **6 карандашей**.
При пяти карандашах ещё возможен случай, когда все они синие,
а шестой карандаш гарантированно будет красным.

## 3. Встреча двух велосипедистов

**Условие:**

Из двух городов навстречу друг другу одновременно выехали два велосипедиста.
Скорость первого — **12 км/ч**, а скорость второго — **15 км/ч**.
Через сколько часов они встретятся, если изначально расстояние между городами было **81 км**?

### Математическая модель

Скорость сближения равна сумме скоростей велосипедистов:

`12 + 15 = 27 км/ч`.

Расстояние до встречи определяется формулой `distance = speed * time`.

```python
distance, speed_first, speed_second, time = Reals(
    "distance speed_first speed_second time"
)

solver_bikes = Solver()
solver_bikes.add(
    distance == 81,
    speed_first == 12,
    speed_second == 15,
    time >= 0,
    distance == (speed_first + speed_second) * time
)

if solver_bikes.check() == sat:
    model_bikes = solver_bikes.model()
    print("Время до встречи:", model_bikes[time], "ч.")
else:
    print("Решение не найдено")
```

Ожидаемый результат: велосипедисты встретятся через **3 часа**,
поскольку `81 / (12 + 15) = 3`.
