```python
from z3 import *

def outcome(theorem, timeout_ms=None):
    solver = Solver()
    if timeout_ms is not None:
        solver.set(timeout=timeout_ms)
    solver.add(Not(theorem))
    result = solver.check()
    if result == unsat:
        return "контрпример невозможен, значит теорема доказана в выбранной теории;"
    if result == sat:
        return f"найден контрпример: {solver.model()}"
    return f"unknown: {solver.reason_unknown()}"

def show(theorem):
    print(outcome(theorem))
```

### 1. Сумма трёх последовательных чисел

Докажите, что сумма любых трёх последовательных целых чисел делится на 3.

```python
n = Int("three_consecutive_n")
theorem = (n + (n + 1) + (n + 2)) % 3 == 0
show(theorem)
```

### 2. Произведение соседних чисел

Докажите, что произведение любых двух соседних целых чисел чётно.

```python
n = Int("neighbors_n")
theorem = n * (n + 1) % 2 == 0
show(theorem)
```

### 3. Разность соседних квадратов

Докажите, что разность квадратов двух соседних целых чисел равна нечётному числу 2n + 1.

```python
n = Int("squares_n")
theorem = (n + 1)**2 - n**2 == 2*n + 1
show(theorem)
```

### 4. Транзитивность порядка

Докажите: если a ≤ b и b ≤ c, то a ≤ c.

```python
a, b, c = Ints("order_a order_b order_c")
theorem = Implies(And(a <= b, b <= c), a <= c)
show(theorem)
```

### 5. Остаток при делении на 3

Целое число от 1 до 10 при делении на 3 даёт остаток 2. Докажите, что это число равно 2, 5 или 8.

```python
n = Int("remainder_n")
theorem = Implies(
    And(1 <= n, n <= 10, n % 3 == 2),
    Or(n == 2, n == 5, n == 8),
)
show(theorem)
```
